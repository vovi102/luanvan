# BigQuery Pilot EDA

Load the T1.2 pilot CSV exports and inspect schema, nulls, numeric ranges, and known Ethereum edge cases.

```python
from pathlib import Path

import pandas as pd

pilot_dir = Path("data/raw/pilot")
if not pilot_dir.exists():
    pilot_dir = Path("../data/raw/pilot")
files = {
    "transactions": pilot_dir / "transactions_pilot.csv",
    "blocks": pilot_dir / "blocks_pilot.csv",
    "token_transfers": pilot_dir / "token_transfers_pilot.csv",
    "contracts": pilot_dir / "contracts_pilot.csv",
}
frames = {name: pd.read_csv(path) for name, path in files.items()}
{name: frame.shape for name, frame in frames.items()}
```

```python
for name, frame in frames.items():
    print(f"\n## {name}")
    frame.info()
    display(frame.head())
```

```python
for name, frame in frames.items():
    print(f"\n## {name} null counts")
    display(frame.isnull().sum().sort_values(ascending=False))
```

```python
tx = frames["transactions"]
display(tx.describe(include="all"))
display(tx[tx["to_address"].isnull()].head())
display(tx[tx["value"].astype(str) == "0"].head())
display(tx[tx["gas_price"].astype(str) == "0"].head())
display(tx[tx["transaction_type"].astype(str) != "0"].head())
```
