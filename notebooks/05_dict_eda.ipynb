# T2.2 Entity Dictionary EDA

Offline EDA over committed dictionary artifacts.


```python
import json
from collections import Counter
from pathlib import Path

base = Path("../src/nl2sparql/linking/dictionary")
entities = json.loads((base / "entities.json").read_text())
aliases = json.loads((base / "aliases.json").read_text())
concepts = json.loads((base / "concepts.json").read_text())

{"entities": len(entities), "aliases": len(aliases), "concepts": len(concepts)}
```

```python
Counter(entry["category"] for entry in entities).most_common()
```

```python
Counter(entry["owner"] for entry in entities).most_common(20)
```
