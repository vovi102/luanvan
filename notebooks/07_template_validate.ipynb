```python
import json
from pathlib import Path

templates_path = Path("../src/nl2sparql/dataset/templates/templates.json")
templates = json.loads(templates_path.read_text(encoding="utf-8"))
len(templates), templates[0]["id"]
```

```python
rendered = []
for template in templates:
    query = template["sparql_template"].format(**template["example_fill"])
    question = template["nl_seed"].format(**template["example_fill"])
    rendered.append((template["id"], question, query))
rendered[:2]
```
